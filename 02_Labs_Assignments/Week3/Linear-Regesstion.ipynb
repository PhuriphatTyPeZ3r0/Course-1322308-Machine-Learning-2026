```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 50
d = 2
X = np.random.rand(N)
Y = 3 * X + 1 + 0.5 * np.random.rand(N)
plt.plot(X, Y, '.r')
```

$R = YX^T(XX^T)^{-1}$

```python
X_ = np.vstack((X, np.ones(N)))
```

```python
R = Y @ X_.T @ np.linalg.inv(X_ @ X_.T)
```

```python
R
```

```python
i = np.array([0, 1])
j = R[0] * i + R[1]
plt.plot(X, Y, '.r')
plt.plot(i, j, 'g')
```

```python
X = np.random.rand(N)
Y = 3 * np.sin(8* X + 0.5) + 0.5*np.random.rand(N)
plt.plot(X, Y, '.r')
```

```python
X_ = np.vstack((X, np.ones(N)))
X__ = np.vstack((X**2, X_))
R = Y @ X__.T @ np.linalg.inv(X__ @ X__.T)

i = np.arange(0, 1, 0.01)
j = R[0] * i**2 + R[1] * i + R[2]
plt.plot(X, Y, '.r')
plt.plot(i, j, 'g')
```

```python
X_ = np.vstack((X**3, X**2, X, np.ones(N)))
R = Y @ X_.T @ np.linalg.inv(X_ @ X_.T)

i = np.arange(0, 1, 0.01)
j = R[0] * i**3 + R[1] * i**2 + R[2] * i + R[3]
plt.plot(X, Y, '.r')
plt.plot(i, j, 'g')
```

```python
X_ = np.vstack((X**4, X**3, X**2, X, np.ones(N)))
R = Y @ X_.T @ np.linalg.inv(X_ @ X_.T)

i = np.arange(0, 1, 0.01)
j = R[0] * i**4 + R[1] * i**3 + R[2] * i**2 + R[3] * i + R[4]
plt.plot(X, Y, '.r')
plt.plot(i, j, 'g')
```

```python
for M in range(2, 15):
  X_ = np.vstack((X, np.ones(len(X))))
  for n in range(2, M + 1):
    X_ = np.vstack((X**n, X_))

  R = Y @ X_.T @ np.linalg.inv(X_ @ X_.T)

  # i = np.arange(0, 1, 0.01)
  # j = 0
  # for n in range(M + 1):
  #   j += R[M - n] * i ** n
  # plt.plot(X, Y, '.r')
  # plt.plot(i, j, 'g')
  # plt.show()

  # error
  j = 0
  for n in range(M + 1):
    j += R[M - n] * X ** n
  e = np.mean(np.abs(Y - j)) # MAE
  print('Degree=', M, 'error=', e)
```

# KNN Regresstion

```python
i = np.arange(0, 1, 0.01)
j = []
K = 5
for i_ in np.arange(0, 1, 0.01):
  d = np.abs(X - i_)

  id = np.argsort(d)[:K]
    
  j.append(np.mean(Y[id]))

plt.plot(X, Y, '.r')
plt.plot(i, j, 'g')
plt.show()
```
